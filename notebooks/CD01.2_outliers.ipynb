```python
# Importación de librerías y supresión de advertencias
import pandas as pd
import warnings
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Lectura de los datos
df = pd.read_csv('../data/raw/bank-additional-full.csv', sep=';')
```

```python
# Se complementa el análsis estadistico con los graficos bloxsplot para las variables numéricas
variables_con_outliers = []

for col in df.select_dtypes(include=['float64', 'int64']).columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1 # rango intercuartilico

    # Se consideran 1.5 para outliers leves
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = df[(df[col] < limite_inferior) | (df[col] > limite_superior)]

    if not outliers.empty:
        print(f'{col}: {len(outliers)} outliers encontrados')
        if col != 'y':
            variables_con_outliers.append(col)

# Crea gráficos ordenados en columnas de acuerdo a la cantidad de variables con outliers
num_columnas = len(variables_con_outliers)
columnas = 3  # Número de columnas por fila
filas = (num_columnas + columnas - 1) // columnas  # Cálculo del número de filas para graficar

fig, ejes = plt.subplots(filas, columnas, figsize=(6 * columnas, 4 * filas))
ejes = ejes.flatten() 

for i, col in enumerate(variables_con_outliers):
    sns.boxplot(x=df[col], ax=ejes[i])
    ejes[i].set_title(f'Outliers en "{col}"')
    ejes[i].set_xlabel(col)

# Ocultar ejes sobrantes si hay menos gráficos que subplots
for j in range(i + 1, len(ejes)):
    ejes[j].axis('off')

plt.tight_layout()
plt.show()
```
